```python
# log1 (interactive - click legend entries to toggle lines on/off,
# double-click to isolate a single line)
fig = go.Figure()
for col in ["gps_altitude", "altimeter_1_altitude", "altimeter_2_altitude"]:
    fig.add_trace(go.Scatter(x=log1_df["timestamp"], y=log1_df[col], mode="lines", name=col))
fig.update_layout(xaxis_title="timestamp", yaxis_title="altitude", title="log1")
fig.show()
```

```python
# log2 (interactive - click legend entries to toggle lines on/off,
# double-click to isolate a single line)
fig = go.Figure()
for col in ["gps_altitude", "altimeter_1_altitude", "altimeter_2_altitude"]:
    fig.add_trace(go.Scatter(x=log2_df["timestamp"], y=log2_df[col], mode="lines", name=col))
fig.update_layout(xaxis_title="timestamp", yaxis_title="altitude", title="log2")
fig.show()
```
